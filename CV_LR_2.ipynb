## <font color='pink'> Функция для считывания изображения по URL картинки

```python
def download_and_read_image(url: str) -> np.ndarray:
    """
    Загрузка изображения по URL и его чтение через OpenCV.

    Вход:
        url (str): URL изображения.

    Выход:
        np.ndarray: Изображение в формате BGR (OpenCV).
    """
    # Скачать изображение
    response = requests.get(url)
    response.raise_for_status()  # Проверка на ошибки при загрузке

    # Преобразовать содержимое в массив numpy
    image_array = np.frombuffer(response.content, np.uint8)

    # Прочитать изображение в формате OpenCV
    image = cv2.imdecode(image_array, cv2.IMREAD_GRAYSCALE)
    if image is None:
        raise ValueError("Не удалось считать изображение.")
    return image
```

## <font color='aquamarine'> 1. Выравнивание яркости

```python
import numpy as np

def align_brightness(image1: np.ndarray, image2: np.ndarray) -> np.ndarray:
    """
    Выравнивает яркость двух изображений на основе нормализации среднего и стандартного отклонения.

    Args:
        image1 (np.ndarray): Первое изображение (эталонное).
        image2 (np.ndarray): Второе изображение.

    Returns:
        np.ndarray: Второе изображение после выравнивания яркости.
    """
    mean1, std1 = np.mean(image1), np.std(image1)
    mean2, std2 = np.mean(image2), np.std(image2)

    aligned_image = (image2 - mean2) * (std1 / std2) + mean1
    return np.clip(aligned_image, 0, 255).astype(np.uint8)
```

## <font color='aquamarine'> 2. Обнаружение ключевых точек и описание дескрипторов

### <font color='pink'> Ключевые точки - (углы Харриса)

```python
def conv2d_manual(image: np.ndarray, kernel: np.ndarray) -> np.ndarray:
    """
    Реализует операцию свертки вручную.

    Args:
        image (np.ndarray): Входное изображение.
        kernel (np.ndarray): Ядро свертки.

    Returns:
        np.ndarray: Результат свертки.
    """
    kernel_height, kernel_width = kernel.shape
    padding_y = kernel_height // 2
    padding_x = kernel_width // 2

    padded_image = np.pad(image, ((padding_y, padding_y), (padding_x, padding_x)), mode='constant')

    result = np.zeros_like(image)

    for i in range(image.shape[0]):
        for j in range(image.shape[1]):
            result[i, j] = np.sum(padded_image[i:i + kernel_height, j:j + kernel_width] * kernel)

    return result
```

```python
def harris_corner_detection(
        image: np.ndarray,
        k: float = 0.05,
        threshold: float = 2e6  # 2.5e5
        ) -> list:
    """
    Реализует оператор Харриса для обнаружения углов.

    Args:
        image (np.ndarray): Градации серого (2D массив).
        k (float): Параметр уравнения Харриса.
        threshold (float): Пороговое значение для выбора углов.

    Returns:
        list: Список объектов cv2.KeyPoint, представляющих углы.
    """
    # Вычисление градиентов
    Ix = np.gradient(image, axis=1)  # Градиент по x
    Iy = np.gradient(image, axis=0)  # Градиент по y

    # Вычисление компонентов матрицы M
    Ixx = Ix ** 2
    Iyy = Iy ** 2
    Ixy = Ix * Iy

    # Применение фильтра Гаусса для сглаживания (прямое свёртывание)
    kernel_size = 3
    sigma = 3
    ax = np.linspace(-(kernel_size // 2), kernel_size // 2, kernel_size)
    gauss = np.exp(-0.5 * np.square(ax) / np.square(sigma))
    kernel = np.outer(gauss, gauss)
    kernel /= np.sum(kernel)


    Sxx = conv2d_manual(Ixx, kernel)
    Syy = conv2d_manual(Iyy, kernel)
    Sxy = conv2d_manual(Ixy, kernel)

    # Оператор Харриса
    det_M = Sxx * Syy - Sxy ** 2
    trace_M = Sxx + Syy # lambda_1, lambda_2 ## eigenvals
    R = det_M - k * (trace_M ** 2)
    print(R.max(), R.min())
    # Выбор углов
    keypoints_coords = np.argwhere(R > threshold)

    # Преобразование координат углов в формат cv2.KeyPoint
    keypoints = []
    for y, x in keypoints_coords:
        # Создание объекта cv2.KeyPoint с фиксированным размером и углом
        keypoints.append(cv2.KeyPoint(x=float(x), y=float(y), size=1, angle=0))

    return keypoints
```

### <font color='pink'> Дескрипотры SIFT

```python
def compute_sift_descriptors(
    image: np.ndarray,
    keypoints: list,
    patch_size: int = 16,
    num_bins: int = 8
    ) -> np.ndarray:
    """
    Расчет SIFT дескрипторов длят очек интереса на изображении.

    Parameters:
        image (np.ndarray): Grayscale изображение (H x W).
        keypoints (list): List cv2.KeyPoint.
        patch_size (int): Размер квадратной области вокруг каждой точки (16).
        num_bins (int): Количество гистограмм для ориентаций (8).

    Returns:
        np.ndarray: Дескриптор размера (N x 128), где N число точек.
    """
    # Ensure the image is in float format
    image = image.astype(np.float32)

    # Compute gradients and orientation using OpenCV's Sobel
    dx = cv2.Sobel(image, cv2.CV_32F, 1, 0, ksize=3)
    dy = cv2.Sobel(image, cv2.CV_32F, 0, 1, ksize=3)

    magnitude = np.sqrt(dx ** 2 + dy ** 2)
    orientation = (np.arctan2(dy, dx) * 180 / np.pi) % 360

    descriptors = []

    # Process each keypoint
    for keypoint in keypoints:
        # Extract coordinates from the KeyPoint object
        x, y = keypoint.pt
        x, y = int(x), int(y)
        half_size = patch_size // 2

        # Check if the patch is within bounds
        if (x - half_size < 0 or x + half_size >= image.shape[1] or
            y - half_size < 0 or y + half_size >= image.shape[0]):
            descriptors.append(np.zeros(128))
            continue

        patch_magnitude = magnitude[y - half_size:y + half_size, x - half_size:x + half_size]
        patch_orientation = orientation[y - half_size:y + half_size, x - half_size:x + half_size]

        # Divide the patch into 4x4 cells
        cell_size = patch_size // 4
        descriptor = []
        for i in range(4):
            for j in range(4):
                cell_magnitude = patch_magnitude[i * cell_size:(i + 1) * cell_size, j * cell_size:(j + 1) * cell_size]
                cell_orientation = patch_orientation[i * cell_size:(i + 1) * cell_size, j * cell_size:(j + 1) * cell_size]

                # Compute histogram of gradients
                hist, _ = np.histogram(cell_orientation, bins=num_bins, range=(0, 360), weights=cell_magnitude)
                descriptor.extend(hist)

        # Normalize the descriptor
        descriptor = np.array(descriptor)
        descriptor /= np.linalg.norm(descriptor) + 1e-8
        descriptor = np.clip(descriptor, 0, 0.2)
        descriptor /= np.linalg.norm(descriptor) + 1e-8

        descriptors.append(descriptor)

    return np.array(descriptors)
```

## <font color='aquamarine'> 3. Сопоставление ключевых точек

### <font color='pink'> Класс для преобразования типов данных
Нужно для использования стандартных функций из OpenCV для визуализации результата

```python
class DMatch:
    def __init__(self, queryIdx, trainIdx, distance):
        self.queryIdx = queryIdx  # Индекс ключевой точки на изображении 1
        self.trainIdx = trainIdx  # Индекс ключевой точки на изображении 2
        self.distance = distance  # Расстояние между дескрипторами

    def to_cv2(self):
        # Преобразование в cv2.DMatch
        return cv2.DMatch(self.queryIdx, self.trainIdx, self.distance)
```

### <font color='pink'> Функция вычисления расстояния в евклидовом пространстве

```python
def euclidean_distance(desc1: np.ndarray, desc2: np.ndarray) -> float:
    """
    Вычисление евклидова расстояния между двумя дескрипторами.

    Параметры:
        desc1 (np.ndarray): Дескриптор для ключевой точки 1.
        desc2 (np.ndarray): Дескриптор для ключевой точки 2.

    Возвращает:
        float: Евклидово расстояние между двумя дескрипторами.
    """
    return np.linalg.norm(desc1 - desc2)
```

### <font color='pink'> Функция сопоставления ключевых точек по дескрипторам

```python
def match_keypoints(keypoints1, descriptors1: np.ndarray, keypoints2, descriptors2: np.ndarray, ratio_threshold: float = 0.97) -> list:
    """
    Сопоставление ключевых точек между двумя изображениями по дескрипторам без использования cv2.

    Параметры:
        keypoints1 (list): Список ключевых точек первого изображения.
        descriptors1 (np.ndarray): Дескрипторы ключевых точек первого изображения.
        keypoints2 (list): Список ключевых точек второго изображения.
        descriptors2 (np.ndarray): Дескрипторы ключевых точек второго изображения.
        ratio_threshold (float): Порог для фильтрации матчей по методу Лиона (default: 0.75).

    Возвращает:
        list: Список матчей в формате [cv2.DMatch(...), ...].
    """

    matches = []

    # Для каждого дескриптора из первого изображения находим два ближайших соседа из второго изображения
    for i, desc1 in enumerate(descriptors1):
        # Вычисление расстояний между текущим дескриптором из первого изображения и всеми дескрипторами второго изображения
        distances = [euclidean_distance(desc1, desc2) for desc2 in descriptors2]

        # Находим два ближайших соседа
        sorted_indices = np.argsort(distances)
        best_idx, second_best_idx = sorted_indices[0], sorted_indices[1]
        best_distance, second_best_distance = distances[best_idx], distances[second_best_idx]

        # Применяем тест Лиона для фильтрации ложных совпадений
        if best_distance < ratio_threshold * second_best_distance:
            match_ = DMatch(queryIdx=i, trainIdx=best_idx, distance=best_distance)
            matches.append(match_)

    # Преобразование кастомных матчей в cv2.DMatch
    return [match_.to_cv2() for match_ in matches]
```

## <font color='aquamarine'> 4. Построение модели преобразования и траектории

### <font color='pink'> Построение модели преобразования

```python
def estimate_transformation(matches, keypoints1, keypoints2):
    """
    Оценивает трансформацию (поворот и сдвиг) между двумя наборами точек.

    Args:
        matches (list): Список объектов cv2.DMatch, представляющих соответствия между ключевыми точками.
        keypoints1 (list): Ключевые точки первого изображения.
        keypoints2 (list): Ключевые точки второго изображения.

    Returns:
        np.ndarray: Матрица 2x3 преобразования.
    """
    # Извлечение координат из keypoints
    points1 = np.array([keypoints1[m.queryIdx].pt for m in matches])
    points2 = np.array([keypoints2[m.trainIdx].pt for m in matches])

    # Средние значения
    centroid1 = np.mean(points1, axis=0)
    centroid2 = np.mean(points2, axis=0)

    # Центрирование
    centered1 = points1 - centroid1
    centered2 = points2 - centroid2

    # Оценка матрицы поворота
    H = np.dot(centered1.T, centered2)
    U, _, Vt = np.linalg.svd(H)
    R = np.dot(Vt.T, U.T)

    # Сдвиг
    t = centroid2 - np.dot(centroid1, R)

    # Построение матрицы 2x3
    transformation = np.hstack([R, t.reshape(-1, 1)])
    return transformation
```

### <font color='pink'> Построение траектории

```python
def compute_trajectory(transformations):
    """
    Построение траектории движения камеры на основе аффинных преобразований
    между последовательными изображениями.

    Параметры:
        keypoints1 (list): Список ключевых точек первого изображения.
        keypoints2 (list): Список ключевых точек второго изображения.
        matches (list): Список объектов cv2.DMatch, представляющих соответствия между ключевыми точками.
        num_images (int): Число изображений (для имитации движения).

    Возвращает:
        list: Траектория движения камеры.
    """
    trajectory = [np.array([0, 0])]
    current_position = np.array([0., 0.])  # Начальная позиция камеры

    for transformation in transformations:

        # Извлечение сдвига из матрицы преобразования
        translation = transformation[:, 2]

        # Обновление позиции камеры
        current_position += translation

        trajectory.append(-current_position.copy())

    return np.array(trajectory)
```

```python
"""
images = [img1, img2, ..., imgN] (list)
"""
# Инициализация данных
prev_image = images[0]
transformations = []

for next_image in images[1:]:

    # Выравнивание яркости
    next_image = align_brightness(prev_image, next_image)

    # Обнаружение ключевых точек
    keypoints1 = harris_corner_detection(prev_image, k=0.04, threshold=2e6)
    keypoints2 = harris_corner_detection(next_image, k=0.04, threshold=2e6)

    # Дескрипторы
    descriptors1 = compute_sift_descriptors(prev_image, keypoints1, patch_size=64, num_bins=8)
    descriptors2 = compute_sift_descriptors(next_image, keypoints2, patch_size=64, num_bins=8)

    # Сопоставление ключевых точек
    matches = match_keypoints(keypoints1, descriptors1, keypoints2, descriptors2)

    # Визуализация текущей пары
    match_image = cv2.drawMatches(
        prev_image, keypoints1,
        next_image, keypoints2,
        matches, None)

    cv2_imshow(match_image)

    transformation = estimate_transformation(matches, keypoints1, keypoints2)
    transformations.append(transformation)

    prev_image = next_image
```
